```python
from sklearn.feature_extraction.text import CountVectorizer, ENGLISH_STOP_WORDS

df["topics_text"] = df["topics"].apply(" ".join)

df["short_text"] = (
    df["name"].fillna("") + " " +
    df["description"].fillna("") + " " +
    df["topics_text"]
).str.lower()

stopwords = list(ENGLISH_STOP_WORDS | {
    "python", "project", "projects", "tool", "tools",
    "github", "code", "using", "based", "2026"
})

vectorizer = CountVectorizer(stop_words=stopwords, max_features=3000)
X = vectorizer.fit_transform(df["short_text"])

word_counts = pd.Series(
    np.asarray(X.sum(axis=0)).ravel(),
    index=vectorizer.get_feature_names_out()
).sort_values(ascending=False)

display(word_counts.head(30))
```
